```python
import functions
import numpy as np
```

```python
theta_input = 2.0           # Twist angle in degrees
delta_input = 0.18          # Lattice mismatch e.g. a_hBN = (1 + delta) a_graphene
tol_theta = 1.0             # Allowed angle tolerance (in degrees)
tol_delta = 0.05           # Allowed mismatch tolerance (fractional)
Umax = 15                   # M,N,M',N' search bound (increase especially when decreasing delta tolerance)

matches = functions.supercell_search(theta_input, delta_input, tol_theta, tol_delta, Umax=Umax)

print("Number of matches found:", len(matches))
print("----------------------------------------")
if matches:
    min_area = min(m['area'] for m in matches)
    min_matches = [m for m in matches if m['area'] == min_area]
    print(f"\nPrinting minimum-area supercells (area factor={min_area}, atom count={4*min_area}):")
    for i, m in enumerate(min_matches, 1):
        print(f"{i}) M={m['M']:3d}, N={m['N']:3d} | Mp={m['Mp']:3d}, Np={m['Np']:3d} | "
              f"θ_eff={m['theta_eff_deg']:.3f}° δ_eff={m['delta_eff']:.5f} area factor={m['area']} atoms={m['atom_count']}")
else:
    print("No matching supercells found within tolerances.")
```

```python
theta = 1.945
delta = 0.21499
M, N, Mp, Np = -6, 1, -5, 1

a = 2.46
a1, a2 = functions.graphene_lattice_vectors(a)
T1, T2 = functions.supercell_vectors(M, N, a1, a2)
T1p, T2p = functions.supercell_vectors_layer2per(Mp, Np, a1, a2, theta, delta)

print("Supercell vectors from layer 1:")
print("T1 =", T1)
print("T2 =", T2)
print("----------------------------------------")
print("Supercell vectors from layer 2:")
print("T1p =", T1p)
print("T2p =", T2p)

print("Check length and angle match:")
print("----------------------------------------")
print("Layer 1:")
print("Lengths: |T1| =", np.linalg.norm(T1), " |T2| =", np.linalg.norm(T2))
print("Angle b/w T1 and T2:", functions.angle_deg(T1, T2), "deg")
print("Layer 2:")
print("Lengths: |T1p| =", np.linalg.norm(T1p), " |T2p| =", np.linalg.norm(T2p))
print("Angle b/w T1p and T2p:", functions.angle_deg(T1p, T2p), "deg")
print("----------------------------------------")
tauA = functions.offset_A_sublattice()
tauB = functions.offset_B_sublattice(a1, a2)
_, fracA = functions.points_in_moire_cell(a1, a2, tauA, T1, T2)
_, fracB = functions.points_in_moire_cell(a1, a2, tauB, T1, T2)

a12 = functions.rotate_and_scale(a1, theta, delta)
a22 = functions.rotate_and_scale(a2, theta, delta)
tauAp = functions.offset_A_sublattice()
tauBp = functions.offset_B_sublattice(a12, a22)
_, fracA2 = functions.points_in_moire_cell(a12, a22, tauAp, T1p, T2p)
_, fracB2 = functions.points_in_moire_cell(a12, a22, tauBp, T1p, T2p)

print(f"Layer 1: {fracA.shape[0]} A atoms, {fracB.shape[0]} B atoms")
print(f"Layer 2: {fracA2.shape[0]} A atoms, {fracB2.shape[0]} B atoms")
print(f"Total atoms: {fracA.shape[0] + fracB.shape[0] + fracA2.shape[0] + fracB2.shape[0]}")
```

```python

functions.write_poscar(f"POSCAR_{theta}_{delta}.vasp", T1, T2, fracA, fracB, fracA2, fracB2, species=("C", "C", "B", "N"), c=25.0, h=3.35)
```
